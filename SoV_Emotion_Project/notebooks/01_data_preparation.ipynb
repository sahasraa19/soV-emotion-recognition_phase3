```python
# Setup: find the repo root so `from src import ...` works (run this notebook from inside the repo).
import sys
from pathlib import Path
ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "src").is_dir()), None)
assert ROOT is not None, "Run from inside the repo (the folder containing src/). In Colab: upload the repo zip, unzip it and %cd into it."
sys.path.insert(0, str(ROOT))
from src import config, utils
print("repo root:", ROOT)
```

```python
# Option A (Colab): upload a zip of photos. Option B (local): copy the photos into data/raw/ yourself and skip this cell.
import zipfile
try:
    from google.colab import files
    up = files.upload()
    config.RAW_DIR.mkdir(parents=True, exist_ok=True)
    for name, content in up.items():
        p = config.RAW_DIR / name
        p.write_bytes(content)
        if name.lower().endswith(".zip"):
            zipfile.ZipFile(p).extractall(config.RAW_DIR / (Path(name).stem + "_unzipped"))
except ImportError:
    print("Not running in Colab - put your photos in", config.RAW_DIR)
```
