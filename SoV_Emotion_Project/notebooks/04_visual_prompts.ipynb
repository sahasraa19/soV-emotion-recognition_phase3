# 04 - Visual prompts and the question we ask
**Stage:** from one image and the same face detections we draw the 10 visual prompts. Only the drawing changes.

| ID | Drawing | Role |
|---|---|---|
| V0 | nothing | baseline |
| V1 | red circle on each face + rest blurred | RedCircle-style (paper Table 2) |
| V2 | box on each face + rest blurred | ReCLIP-style (paper Table 2) |
| V3 | box | ablation: Box |
| V4 | box + number | ablation: Box + Number |
| V5 | box + number + 5 landmark dots | full SoV |
| V6 | number tag only | ablation: ID alone |
| V7 | 5 landmark dots only | ablation: landmarks alone |
| V8 | full SoV with **small** dots | Phase 3: tests whether big dots hid the faces |
| V9 | landmarks only, **small** dots | Phase 3 |

| | |
|---|---|
| **Input** | `data/dataset.json` (notebook 03) |
| **Output** | `results/prompt_variants.png`; the text question comes from `src/prompts.py` |
| **Code used** | `src/prompts.py` |

```python
# Setup: find the repo root so `from src import ...` works (run this notebook from inside the repo).
import sys
from pathlib import Path
ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "src").is_dir()), None)
assert ROOT is not None, "Run from inside the repo (the folder containing src/). In Colab: upload the repo zip, unzip it and %cd into it."
sys.path.insert(0, str(ROOT))
from src import config, utils
print("repo root:", ROOT)
```

```python
import cv2, matplotlib.pyplot as plt
from src import prompts
DATA = utils.load_json(config.DATASET_JSON)
rec = max(DATA, key=lambda r: r["n_faces"])
fig, axes = plt.subplots(2, 5, figsize=(34, 11))
for ax, v in zip(axes.ravel(), config.ALL_VARIANTS):
    ax.imshow(cv2.cvtColor(prompts.render_record(rec, v), cv2.COLOR_BGR2RGB)); ax.set_title(config.VARIANT_TITLES[v], fontsize=14); ax.axis("off")
plt.tight_layout()
config.RESULTS_DIR.mkdir(exist_ok=True)
plt.savefig(config.RESULTS_DIR / "prompt_variants.png", dpi=90); plt.show()
print("Check: numbers and boxes sit on the right faces and landmarks are visible. If not, fix the labels/detections in notebooks 02-03.")
```
